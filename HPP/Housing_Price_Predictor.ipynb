```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv("Housing.csv")

train, test = train_test_split(df, test_size = 0.2, random_state = 42)
```

```python
train.head()
```

```python
train["mainroad"] = train["mainroad"].map({"no": 0, "yes": 1})
train["guestroom"] = train["guestroom"].map({"no": 0, "yes": 1})
train["basement"] = train["basement"].map({"no": 0, "yes": 1})
train["hotwaterheating"] = train["hotwaterheating"].map({"no": 0, "yes": 1})
train["airconditioning"] = train["airconditioning"].map({"no": 0, "yes": 1})
train["prefarea"] = train["prefarea"].map({"no": 0, "yes": 1})
train["furnishingstatus"] = train["furnishingstatus"].map({"unfurnished": 0, "semi-furnished": 1, "furnished": 2})
train.head()
```

```python
test["mainroad"] = test["mainroad"].map({"no": 0, "yes": 1})
test["guestroom"] = test["guestroom"].map({"no": 0, "yes": 1})
test["basement"] = test["basement"].map({"no": 0, "yes": 1})
test["hotwaterheating"] = test["hotwaterheating"].map({"no": 0, "yes": 1})
test["airconditioning"] = test["airconditioning"].map({"no": 0, "yes": 1})
test["prefarea"] = test["prefarea"].map({"no": 0, "yes": 1})
test["furnishingstatus"] = test["furnishingstatus"].map({"unfurnished": 0, "semi-furnished": 1, "furnished": 2})
```

```python
X_train = train.drop("price", axis = 1)
y_train = train["price"]

X_test = test.drop("price", axis = 1)
y_test = test["price"]
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)
```

```python
print("MAE:", mae)
print("RMSE:", rmse)
print("R²:", r2)
```
